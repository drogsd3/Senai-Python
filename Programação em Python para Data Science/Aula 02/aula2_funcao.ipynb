## Escrevam uma função que retorne o maior valor de dois números

máximo(5,6) resultado deve ser o 6
maximo(2,1) resultado deve ser o 2

```python
def maximo(a,b):
    if a > b:
        print(f'O numero maior numero é: {a}')
    elif b > a:
        print(f'O numero maior numero é: {b}')
    else:
        print('O dois numeros são iguais')
```

```python
maximo(5,6)
maximo(2,1)
maximo(1,1)
```

- Escreva uma funçao que receba a base e a altura de um triangulo e retorne a sua area ( A = Base X Altura/2)
- Escreva uma função que calcule a area de um retangulo ( A = Lado X Altura)
- Escreva uma função que calcule a area de um trapezio ( A = ((Base Maior + Base Menor) X Altura)/2)
- Escreva uma função que calcule a area de um circulo ( A = PI * Raio²)

```python
%%writefile geometria.py
from math import pi
def AreaTriangulo(base,altura):
    area = base * altura /2
    return area

def AreaRetangulo(lado, altura):
    area = lado * altura
    return area

def AreaTrapezio(baseMaior,baseMenor, altura):
    area = ((baseMaior+baseMenor)*altura)/2
    return area

def AreaCirculo(raio):
    area = pi * raio**2
    return area 
```

```python
import geometria as geo

while True:
    opcao= int(input("""
Qual calculo você quer fazer?
1 - Area do triangulo.
2 - Area do retangulo.
3 - Area do trapezio.
4 - Area do circulo.
0 - Sair"""))
    if opcao == 1:
        base = float(input('Informe a base desse triangulo: '))
        altura = float(input('Informe a altura: '))
        print(f'\nA area desse triangulo é: {geo.AreaTriangulo(base,altura)}')
    elif opcao == 2:
        lado = float(input('Informe o valor do lado: '))
        altura = float(input('Informe a altura: '))
        print(f'\nA area desse retangulo é: {geo.AreaRetangulo(lado,altura)}')
    elif opcao == 3:
        baseMaior = float(input('Informa o valor da base maior:'))
        baseMenor = float(input('Informa o valor da base menor:'))
        altura = float(input('Informe a altura: '))
        print(f'\nA area desse trapezio é: {geo.AreaTrapezio(baseMaior,baseMenor,altura)}')
    elif opcao == 4:
        raio = float(input('Digite o valor do raio: '))
        print(f'\nA area desse circulo é: {geo.AreaCirculo(raio):.2f}')
    elif opcao == 0:
        print("\nTCHAU")
        break
    else:
        print('Por Favor informe um valor valido')
```
