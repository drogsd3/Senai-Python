# Indexação em Strings

Sabemos que as strings são uma sequencia , o que significa que o python usa indices para chamar partes da sequencia. Vamos Fazer um teste

Em python usamos colchetes [] após um objeto para chamar seu indice. Devemos também notar que a indexação começa em 0. Vamos criar uma variavel "S" e testar

```python
# S é uma string
s = "Hello World"
```

```python
s
```

```python
print (s)
```

```python
print (s[0])
print(s[4])
```

- Podemos usar um : dentro do [] para executar um corte (slice) que pega tudo até um, ponto definido. Por exemplo: 

```python
print(s[1:])
print(s[:5])
print(s[:1])
```

```python
print(s[-1])
print(s[:-1])
print(s[:10])
```

```python
print(s[::-1])
```

```python
print(type(s))
s[0] = "A"
print(s)
```

```python
s + ' Seja bem vindo ao senai'
```

```python
s = s + " Seja bem vindo ao senai"
print(s)
```

```python
s = "Alô Mundo"
s[0] = 'a'
print(s)
```

```python
l = list('Alô Mundo')
l[0] = 'a'
print(l)

s = "".join(l)

print(s)
```
