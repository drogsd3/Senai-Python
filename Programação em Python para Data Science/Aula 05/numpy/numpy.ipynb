### Qual a diferença de listas e array NumPy

Lista do Python e arrays do NumPy tem algumas em comum:
- Guardam varios elementos em sequencia
- Permite adicionar, remover, concatenar elementos

Mas os arrays NumPy têm algumas vantagens bem importantes:
- São mais rapidas e eficentes em operações matemáticas
- Usam menos memórias
- Tem varias ferramentas extras para trabalhar com dados numéricos

**Resumindo** Listas são mais flexíveis, mas os arrays do NumPy, são melhores quando o foco são numeros e cálculos

- Operações Matematicas: O array do Numpy, faz conta de uma vez em todos os elementos (vetorial), já lista do python faz item por item (escalar)

```python
lista = [1,2,3]

print("Resultado da Operação multiplicação")
print(lista*2)
```

```python
a = [1,3,5]
b = [5,7,9]
```

```python
valor = []
for i in range(3):
    valor.append(a[i] * b[i])
print(valor)
```

### Criando Array Numpy

- A base do numpy são os array, que guardam dados do mesmo tipo e podem ter varias dimensões.
- Você pode criar arrays a partir de uma lista comum do python

```python
import numpy as np

# Isso é uma lista
lista = [10,20,30,40]

array = np.array(lista)

print(f"Lista: {lista}, tipo: {type(lista)}")
print(f"Array: {array}, tipo: {type(array)}")
```

### Funções Elementares
---
- Soma

```python
array_a = np.array([1,2,3,4,5])
array_b = np.array([10,20,30,40,50])
escalar = 5

#Soma de Arrays
print(f'Soma de Arrays (operador +): {array_a + array_b}')
print(f'Soma de Arrays com variavel (operador +): {array_a + escalar}')
```

---
- Subtração

```python
print(f'Subtração de Arrays (operador -): {array_a - array_b}')
print(f'Subtração de Arrays com variavel (operador -): {array_a - escalar}')
```

---
- Divisão

```python
print(f'Divisão de Arrays (operador /): {array_a / array_b}')
print(f'Divisão de Arrays com variavel (operador /): {array_b / escalar}')
```

---
- Multiplicação

```python
print(f'Multiplicação de Arrays (operador *): {array_a * array_b}')
print(f'Multiplicação de Arrays com variavel (operador *): {array_a * escalar}')
```

----
- Exponenciação

```python
print(f'Exponenciação de Arrays (operador **): {array_b ** array_a}')
print(f'Exponenciação de Arrays com variavel (operador **): {array_a ** escalar}')
```

```python
import numpy as np

m1 = np.array([
    [40,50,60],
    [70, 80, 90]
])

print(m1)
print(f"Formato do Array: {m1.shape}")
print(f"Dimensão do Array: {m1.ndim}")
print(f"Tipo do Array: {m1.dtype}")
```

```python
m2 = np.array([[2,3,4,5],[9,8,7,6],[50,60,70,60]])
print(m2)
print(f"Formato do Array: {m2.shape}")
print(f"Dimensão do Array: {m2.ndim}")
print(f"Tipo do Array: {m2.dtype}")
```

```python
print(m2[1, 1:3])
print(m2[0])
print(m2[2])

print(m2[:,1:3]) #Coluna
```

### Crie um array com os numeros de 1 a 5 e imprima cada elemento usando um loop

```python
numeros = np.array([[1,2,3],[4,5,6]])

a,b = numeros.shape

for i in range(a):
    for y in range(b):
        print(numeros[i,y], end=' ')
    print()
```
