### Numpy no dia a dia: Array 1d

Aplicar em exemplos reais

1. Criando Arrays do Mundo Real

- Imagine que você tem os preços de produtos em uma loja R$ 10.50, R$ 23, R$5.50 , R$ 12, R$7.50

2. Atributos do array (Numero de Dimensões, Formato, Tipos dados)

3. A loja vai aplicar 10% de aumento em todos seus produtos

```python
import numpy as np
```

```python
precos = np.array([10.50,23,5.5,12,7.5])
precos_desconto = np.array([])

print(f'Dimensão do array: {precos.shape}')
print(f'Formato do array: {precos.ndim}')
print(f'Tipo de dados do array: {precos.dtype}')

precos_desconto = precos * 0.9

print(f"Os preços com descontos: {precos_desconto}")
```

### Funções Elementares

- Calcular algumas estatística (Maior Valor, Menor Valor e a Media)

```python
print(f"A lista de preço é: {precos}")
print(f"O maior preço é: R${precos.max()}")
print(f"O menor preço é: R${precos.min()}")
print(f"A media dos preços é: {np.round(precos.mean(),2)}")
```

### Calculos Rapidos

Criar um array de quantidades vendidas [5,2,10,3,7]

```python
quantidade_venda = np.array([5,2,10,3,7])
```

```python
faturamento = precos*quantidade_venda

print(f"O faturamento é: {faturamento} ")
print(f"O total do faturamento é: {faturamento.sum()}")
```

### Criação e Soma de Vetores

1. Crie duas listas (lista_a e lista_b) com 100.000 numeros inteiros.

2. Crie duas listas (lista_c e lista_d) com 100.000 numeros

3. Some os elementos das duas listas(lista_a e lista_b) e armazene na lista "resultado_lista_soma" ( Use for e listas)

4. Converta a lista_a e lista_b para arrays Numpy (array_a e array_b)

5. Converta a lista_c e lista_d para arrays Numpy (array_c e array_d)

6. Use operação vetorizada (+) para somar array_a e array_b (resultado_array_soma)

7. Compare os resultados (resultado_lista_soma e resultado_array_soma) para verificar se são iguais

```python
lista_a = list(range(100000))
lista_b = list(range(100000))
lista_c = list(range(100000))
lista_d = list(range(100000))
resultado_lista_soma = []
resultado_array_soma = np.array([])

for i in range(100000):
    resultado_lista_soma.append(lista_a[i] + lista_b[i])

array_a = np.array(lista_a)
array_b = np.array(lista_b)
array_c = np.array(lista_c)
array_d = np.array(lista_d)

resultado_array_soma = array_a + array_b

if (resultado_array_soma.all() == np.array(resultado_lista_soma).all()):
    print("Ambos arrays são iguais.")
else:
    print("Não são iguais")
```
