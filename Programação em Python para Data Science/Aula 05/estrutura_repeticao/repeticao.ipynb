### Atividade 3: média de consumo de energia

**Contexto** Você tem uma lista de consumo de energia (kwh) de 10 residências. Calcular a média e identificar quais consumiram acima da média

consumo = [250,300,400,220,310,280,270,330,290,150]

- 1 Calcular a Média de Consumo
- 2 Mostrar na tela quais valores estão acima da média


```python
def calcular_media(consumo):
    media = 0 
    for i in consumo:
        media += i
    media /= len(consumo)
    print("A media é ", media)
    return media
```

```python
consumo = [250,300,400,220,310,280,270,330,290,150]

media = calcular_media(consumo)

for i in consumo:
    if i > media:
        print(f'O consumo de {i} Kwh dessa residencia foi maior que a media.')
```

**Contexto** Uma lista contém os valores de vendas diárias. Alguns dias não houve registros e no sistema aparece, apenas "None".

lista = [150,200,None,180,None,220,170]

- Escrever um codigo que: Conte quantos valores ausentes.

```python
lista = [150,200,None,180,None,220,170]
contador = 0
```

```python
def limpar_lista(lista):
    listanova = []
    contador = 0
    for l in lista:
        if l != None:
            listanova.append(l)
        else:
            contador +=1
    return contador,listanova            
```

### Atividade 8: Analise de Sentimentos

**Contexto** Uma lista contém avaliaçãoes de clientes. Classifique cada avaliação como postitiva, negativa ou neutra com base em palavras chaves.

avaliacao = [otimo,pessimo,bom,ruim,excelente,regular,terrivel]

#### Regras:
- Positiva : otimo, bom e excelente.
- Negativa : pessimo, ruim e terrivel.
- Neutro : regular

```python
avaliacao = ['ótimo','pessímo','bom','ruim','excelente','regular','terrivel']
positivo = 0
negativo = 0
neutro = 0
```

```python
import unicodedata
import re

def remover_acentos(texto):
    texto_normalizado = unicodedata.normalize('NFD', texto)
    return re.sub(r'[\u0300-\u036f]', '', texto_normalizado)
```

```python
for a in avaliacao:
    item = remover_acentos(a)
    if item.lower() == 'otimo' or item.lower() == 'bom' or item.lower == 'excelente':
        positivo += 1
    elif item.lower() == 'regular':
        neutro += 1
    else:
        negativo += 1
print(f"""------- Analise de Sentimentos dos Clientes -------
Quantidade de notas:
* Positiva: {positivo}
* Regular: {neutro}
* Negativa: {negativo}""")
```
