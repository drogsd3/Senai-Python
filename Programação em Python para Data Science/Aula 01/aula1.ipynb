- Converta as expressões matematicas para que possam ser calculadas com o Python:

- 10 + 20 * 30
- 4²/30
- (9³ + 2) x 6 - 1

```python
print(f'10 + 20 * 30 = {10+20*30}')
print(f'4²/30 = {4**2/30:.2f}')
print(f'(9³ + 2) x 6 - 1 = {(9**3 + 2) * 6 - 1}')
```
