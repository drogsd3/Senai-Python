````python
import pandas as pd
````

````python
dicionario = {
'Austria' : 9006398,
'Alemanha' : 83289318,
'Bélgica' : 11589623,
'Bulgária' : 7000039,
'Chipre' : 1207361,
'Croácia' : 4076246,
'Dinamarca' : 5806015,
'Eslováquia' : 5459642,
'Eslovênia' : 2095861,
'Espanha' : 47351567,
'Estônia' : 1324820,
'Finlândia' : 5523231,
'França' : 66977107,
'Grécia' : 10423056,
'Hungria' : 9775564,
'Irlanda' : 4882495,
'Itália' : 60233948,
'Letônia' : 1906743,
'Lituânia' : 2793471,
'Luxemburgo' : 613894,
'Malta' : 493559,
'Países Baixos' : 17134872,
'Polônia' : 37974750,
'Portugal' : 10286263,
'República Tcheca' : 10693939,
'Romênia' : 19405156,
'Suécia' : 10106005
}
````

````python
populacao_ue = pd.Series(dicionario)
print(populacao_ue)
````

````python
print(f"A população da França: {populacao_ue.França}")

````

````python
print(f"A lista em ordem alfabetica: {populacao_ue.align}")
````

````python
print(f"Total de paises: {(populacao_ue.size)}")
````

````python
print(f"A media da população é: {populacao_ue.mean():.2f}")
````

````python
print(f"A mediana é: {populacao_ue.median():.2f}")
````

````python
print(f"Desvio Padrão: {populacao_ue.std():.2f}")
````

````python
print(f"Pais com maior população: {populacao_ue.idxmax()} - {populacao_ue.max()}")
````

````python
print(f"População Total: {populacao_ue.sum()}")
````

````python
print(f"Porcentagem população Espanha em relação a UE: {(populacao_ue.Espanha/populacao_ue.sum())*100:.2f}%")
````

````python
#Solução Aula - Professor
paises_mais_populosos = populacao_ue.sort_values(ascending=False)
print(f"--- Paises mais populosos ---\n{paises_mais_populosos.head(5)}")
````

````python
#Solução Aula - Professor
paises_selecionado = populacao_ue.loc[['França','Alemanha','Croácia']]
print(f"Média dos Paises selecionados: {paises_selecionado.mean():.2f}")
````
