```python
import pandas as pd
import numpy as np

dataframe_vazio = pd.DataFrame()

print("Tipo de dado:",type(dataframe_vazio))
```

```python
# Criando DataFrame a partir de dicionarios

dados_dicionario = {
    'Data': ['2020-01-01','2020-02-01','2020-03-01','2020-04-01','2020-05-01'],
    'Temperatura': [29.1,31.2,28.5,28.0,24],
    'Classificação': ['Quente','Muito quente','Quente','Quente','Moderado']
}

print("Tipos de Dados",type(dados_dicionario))
```

```python
dicionario_dataframe = pd.DataFrame(dados_dicionario)

display(dicionario_dataframe)
```

```python
# Criando Arquivo CSV

pd.DataFrame.to_csv(dicionario_dataframe, 'arquivo.csv', sep=',', index=False)
```

```python
# Criar um DataFrame a partir de um CSV

df = pd.read_csv('arquivo.csv')
df
```

### Desafio 2

Dados do Exame

Nome [Ana,Bruno,Camila,Daniel,Fernanda,Gabriel,Helena,Igor,Juliana,Lucas]
pontuação [12.5,9,16.5,None,9,20,14.5,None,8,19]
tentativas [1,3,2,3,2,3,1,1,2,1]
qualificado [sim,não,sim,não,não,sim,sim,não,não,sim]

#### Indices
indice [A,B,C,D,E,F,G,H,I,J]

```python
nomes = ["Ana", "Bruno", "Camila", "Daniel", "Fernanda", "Gabriel", "Helena", "Igor", "Juliana", "Lucas"]
pontuacao = [12.5,9,16.5,None,9,20,14.5,None,8,19]
tentativas = [1,3,2,3,2,3,1,1,2,1]
qualificado = ["sim", "não", "sim", "não", "não", "sim", "sim", "não", "não", "sim"]
indice = ['A','B','C','D','E','F','G','H','I','J']

dicionario_desafio2 = {
    'nome': nomes,
    'pontuação': pontuacao,
    'tentativas': tentativas,
    'qualificado': qualificado
}

dataframe_desafio2 = pd.DataFrame(dicionario_desafio2, index=indice)
display(dataframe_desafio2)

pd.DataFrame.to_excel(dataframe_desafio2,'desafio.xlsx',index=False)
```
