### O que são Pandas
---
**Pandas** é uma poderosa biblioteca de software livre, para a linguagem Python, fundamental para analise e manipulação de dados. Ele oferece estruturas de dados especializadas, otimizadas para realizar operações eficientes com tabelas númericas e series temporais.

O nome "Pandas" tem origem na expressão *Panel Data*, um termo da **Econometria** que se refere a conjuntos de dados que contêm observações sobre multiplos periodo de tempo para as mesmas unidades (como individuo, empresas, países)

### Principais funcionalidades do Pandas
---
O Pandas oferece um conjunto abrangentes de funcionalidades para trabalhar com dados, incluindo:
- **Manipulação de Dados com indexação personalizada**: Permite acessar dados de forma flexivel usando rotulos (índices).
- **Leitura e escrita de dados em diversos formatos**: Suporta a leitura e escrita de dados em ampla variedade de formatos, como CSV, Excel, Json, SQL e outros.
- **Tratamento de dados Ausentes**: Oferece ferramentas para identificar, limpar e lidar com dados faltantes (NaN).
- **Seleção e filtragem avançandas**: Permite dividir, fatiar e criar subconjuntos de dados extensos com base me critérios complexos.
- **Manipulação de Series Temporais**: Inclui funcionalidades especificas para trabalhar com dados indexados por tempo.
- **União e combinação de Dados**: Facilita a junção (merge) e a concatenação de diferentes conjuntos de dados.

### Estrutura de dados no Pandas

A biblioteca pandas se baseoa em duas Estruturas de dados principais

- **Séries (Series)**: Um estrutura de dados unidimensional rotulada. Pode ser pensada como uma coluna de uma tabela, um array numpy unidimensional com rótulos (indices) ou uma coluna de dataframe.
- **Dataframe**: A principal estrutura de dados do Pandas. Respresenta uma tabela bidimensional de dados, onde linhas geramente representam observações e as colunas representam as variaveis.

```python
import pandas as pd
import numpy as np
```

### Series

O primeiro tipo de dados que aprenderemos é a Serie.
A Serie é muito semelhante a uma matriz Numpy, o que diferencia a matriz Numpy de uma Serie, é que uma Serie pode ter Rótulos de eixo, o que significa que pode ser indexado por um rótulo.

### Criando uma serie

Criar Objetos Series vazio invocando o metodo Series()

```python
#Objeto Series Vazio
series = pd.Series()

print(f'Tipo: f{type(series)}')
print(series)
```

Podemos converter para uma serie:
- Lista.
- Numpy array.
- Dicionário.

```python
labels = ["a","b","c"] # Lista com Strings
minha_lista = [10,20,30] # Lista com Inteiros
array = np.array([10,20,30]) # Array Numpy
dicionario = {"a":10, "b":20, "c":30} # Dicionario com chaves e valores
```

```python
pd.Series(minha_lista) # Cria uma série com os dados da lista
```

Com frequencia, precisamos criar uma serie com um indice que identifique cada ponto de dado com um rótulo. Neste caso passamos o paramentro index = [] contendo uma lista de valores que serão usados com indices:

```python
print("Minha lista:",minha_lista)
print("Meus rotulos:",labels)

pd.Series(minha_lista,labels)
pd.Series(data=minha_lista, index=labels)
```

```python
indice_perso = pd.Series([12,34,21,-9,0,7], index=['A','B','C','D','E','F'])
print(indice_perso)
```

```python
pd.Series(array,index=[1,2,3])
```

```python
pd.Series(dicionario)
```
