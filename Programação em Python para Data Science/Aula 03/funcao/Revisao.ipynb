### O professor precisa de um programa para ajudar no seu dia a dia, para calcular a idade dos alunos, usando com base o ano de nascimento. Nosso desafio é fazer uma função que receba o ano de nascimento e o ano atual e retorne a idade correspondente.

- Digite a data de nascimento: 2005
- Digite o ano atual: 2025

~~~python

"A idade é 20 anos"

```python
import datetime

def ano_idade(ano_nascimento):
    data_atual = datetime.date.today()
    ano_atual = data_atual.year
    idade = ano_atual - ano_nascimento
    return idade
```

```python
ano_nascimento = int(input('Digite o ano do seu nascimento: '))
idade = ano_idade(ano_nascimento)
print(f"A idade é {idade} anos")
```
