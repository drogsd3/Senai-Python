```python
notas = [0,0,0,0,0]

#print(len(notas))

soma = 0
x = 0
while x < 5:
    notas[x] = float(input(f'Nota {x+1}: '))
    soma += notas[x]
    x += 1
x = 0
while x < 5:
    print(f'Nota {x+1}: {notas[x]:.2f}')
    x += 1
print(f'Média: {soma/x:.2f}')
```

### Desenvolva um programa que leia duas listas e que gere um terceira, com os elementos das duas primeiras listas.

```python
lista1 = list(range(10))
lista2 = list(range(10, 20))
lista3 = lista1[:] + lista2[:]
#lista3 = []
#lista3.extend(lista1 + lista2)
print(f'A lista Um contém os seguintes elementos {lista1}')
print(f'A lista Dois contém os seguintes elementos {lista2}')
print(f'A lista 3 contém os elementos das ultimas duas {lista3}')
```

```python
l1 = []
l2 = []
x=0
while True:
    n = int(input('Digite um numero (0 - Sair): '))
    if n == 0:
        break
    l1.append(n)
print('Vamos adicionar elementos em outra lista')
while True:
    n = int(input('Digite um numero (0 - Sair): '))
    if n == 0:
        break
    l2.append(n)
print('Agora vamos adicionar as duas listas em uma terceira')

l3 = l1[:]+ l2[:]


print(f'A lista Um contém os seguintes elementos {l1}')
print(f'A lista Dois contém os seguintes elementos {l2}')
print(f'A lista 3 contém os elementos das ultimas duas {l3}')
```
