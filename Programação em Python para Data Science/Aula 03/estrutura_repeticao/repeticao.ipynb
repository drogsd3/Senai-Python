### Atividade 1: filtrando temperaturas

**Contexto** Você recebeu uma lista de temperatura, registrada durante a semana. O nosso objetivo é identificar os dias que foram quentes, considera-se quente acima de 30° Celsius

temperaturas = [28,32,35,29,31,27,33]

```python
temperaturas = [28,32,35,29,31,27,33]

for i in temperaturas:
    if i > 30:
        print(f'O {temperaturas.index(i)+1}° foi um dia Quente com {i}° Celsius')
    else:
        print(f'O {temperaturas.index(i)+1}° foi um dia Ameno com {i}° Celsius')
```

### Atividade 2: Classificando nivel de poluição

**Contexto** Tenho uma lista que contém os niveis de poluição de diferentes bairros de São Paulo.
Classifique cada bairro como "Seguro", "Moderado" ou "Critíco"

Regras: 
- Até 50: Seguro
- Entre 51 e 100: Moderado
- Acima de 100: Critíco

medidas = [45,78,100,60,30,150,85]


```python
medidas = [45,78,100,60,30,150,85]

for i in medidas:
    if i <= 50:
        print(f'Nivel: {i} - Poluição: Seguro')
    elif i <= 100:
        print(f'Nivel: {i} - Poluição: Moderado')
    else:
        print(f'Nivel: {i} - Poluição: Critico') 
```

### Atividade 3: média de consumo de energia

**Contexto** Você tem uma lista de consumo de energia (kwh) de 10 residências. Calcular a média e identificar quais consumiram acima da média

consumo = [250,300,400,220,310,280,270,330,290,150]

- 1 Calcular a Média de Consumo
- 2 Mostrar na tela quais valores estão acima da média


```python
def calcular_media(consumo):
    media = 0 
    for i in consumo:
        media += i
    media /= len(consumo)
    print("A media é ", media)
    return media
```

```python
consumo = [250,300,400,220,310,280,270,330,290,150]

media = calcular_media(consumo)

for i in consumo:
    if i > media:
        print(f'O consumo de {i} Kwh dessa residencia foi maior que a media.')
```

### Atividade 4: limpeza de dados

**Contexto** Uma lista contem dados de idade de participantes de uma pesquisa, mas alguns valores estão incorretos (negativos ou acima de 120)

lista = [25,34,-3,45,130,29,60,-1,90]

```python
lista = [25,34,-3,45,130,29,60,-1,90]

for i in lista:
    if i < 0 or i > 120:
        lista.remove(i)

print(lista)
```

### Atividade 5

**Contexto** Uma lista que contém palavras extraidas de comentario do reclame aqui sobre o SENAI. Conte quantas vezes a palavra "Excelente" aparece.

lista = ["bom","excelente","ruim","EXCELENTE","regular","excelente","bom"]

```python
lista = ["bom","excelente","ruim","EXCELENTE","regular","excelente","bom"]
contador = 0

for i in lista:
    if i.lower() == 'excelente':
        contador += 1

print(f"Nessa lista contem {contador} avaliações 'Excelente'")
```
