### Simulando uma fila de banco


```python
from IPython.display import clear_output
ultimo = 10
fila = list(range(1,ultimo+1))
```

```python
while True:
    #clear_output()
    print(f'\nExistem {len(fila)} clientes na fila')
    print(f'\nFila Atual: {fila}')
    print(f'Digite F para adicionar um cliente ao fim da fila')
    print(f'Ou A para realizar o Atendimento. (S para Sair)')
    operacao = input('Operação (F,A ou S)')

    for f in operacao:
        if f == "A":
            if len(fila) > 0:
                atendido = fila.pop(0)
                print(f'\nCliente {atendido} atendido')
            else:
                print('Fila Vazia!')
        elif f == "F":
            ultimo += 1
            fila.append(ultimo)
        elif f == "S":
            break
        else:
            print('opção invalida')
    if f == "S":
        break
```

Modificar o programa anterior, para trabalhar com duas filas. Para facilitar seu trabalho, considere os comandos
- A para Atender Fila 1
- B para Atender Fila 2
- F para Adicionar Fila 1
- G para Adicionar FIla 2

```python
ultimo = [5,5]
fila1 = list(range(1,ultimo[0]+1))
fila2 = list(range(1,ultimo[1]+1))
rep = True
```

```python
while rep:
    clear_output()
    print(f'\nExistem {len(fila1)} clientes na primeira fila')
    print(f'\nExistem {len(fila2)} clientes na segunda fila')
    print(f'\n1° Fila: {fila1}\n2° Fila: {fila2}')
    print("""
------- ATENDIMENTO ----------
A - Para Atender 1° Fila.
B - Para Atender 2° Fila.
F - Para Adicionar 1° Fila.
G - Para Adicionar 2° Fila.
S - Para Finalizar Programa.""")
    operacao = input('Digite uma das opções: ')

    for f in operacao:
        if f == "A":
            if len(fila1) > 0:
                atendido = fila1.pop(0)
                print(f'\nCliente {atendido} atendido')
            else:
                print('Fila Vazia!')
        elif f == "B":
            if len(fila2) > 0:
                atendido = fila2.pop(0)
                print(f'\nCliente {atendido} atendido')
            else:
                print('Fila Vazia!')
        elif f == "F":
            ultimo[0] += 1
            fila1.append(ultimo[0])
        elif f == "G":
            ultimo[1] += 1
            fila2.append(ultimo[1])
        elif f == "S":
            rep = False
            break
        else:
            print('Opção Invalida')       
```
