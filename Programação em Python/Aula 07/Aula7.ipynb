# Fazer um contador de foguete

```python
import time
print('Lançamento de Foguete Iniciando...')
for i in range(3, 0, -1):
    print(f'Preparando lançamento...{i}')
    time.sleep(1)
print('***LANÇAMENTO INICIADO***')
```

# Fatiamento de String (slice)

```python
# Exemplo utilizando email:
email = 'profcleisson@gmail.com'

# Pegando o sinal de Arroba
print(f'Letra do indice 12: {email[12]}')
print(f'Letra do indice -1: {email[-1]}')


# Pegando intervalo da String
print(email[:12])
# Exibir apenas a palavra prof
print(email[:4])
# Exibir somente o nome cleisson
print(email[4:12])
```
