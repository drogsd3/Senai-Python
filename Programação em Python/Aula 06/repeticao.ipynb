# Estrutura de Repetição While

```python
print(1)
print(2)
print(3)
```

```python
n = 1
while n <= 5:
    print(n)
    n += 1
```

```python
repeticao = int(input('Digite a quantidade de repetições que o programa irá fazer: '))
x = 1
while x <= repeticao:
    print(x)
    x += 2
```

# Fazer um programa para exibir os resultados no formato de tabuada.

```python
x = int(input('Digite qual a tabuada você quer: '))
repeticao = int(input(f'A Tabuada será do {x}, quer finalizar em qual numero: '))
y = 1
while y <= repeticao:
    print(f'{x} X {y} = {x*y}')
    y += 1
```

### Vejamos um programa para corrigir um teste de multipla escolha com 3 questões. A resposta da primeira é "b", da segunda "a" e da terceira "d", o programa conta um ponto para cada resposta correta.

```python
pontos = 0
questão = 1

while questão <= 3:
    resposta = input(f'Resposta da Questão {questão}')
    if questão == 1 and resposta.lower() == "b":
        pontos += 1
    if questão == 2 and resposta.lower() == "a":
        pontos += 1
    if questão == 3 and resposta.lower() == "d":
        pontos += 1
    questão += 1
print(f'O aluno fez {pontos} ponto(s)')
```

# Calcule a media de 4 notas

```python
notas = 0
media = 0
fim = 4
inicio = 1
while inicio <= fim:
    notas += float(input(f'Digite a nota da {inicio}ª nota: '))
    inicio += 1
media = notas / 4
print(f'A media é {media}')
```

```python
notas = []
media = 0
fim = 4
inicio = 0
while inicio < fim:
    notas.append(float(input(f'Digite a nota da {inicio}° nota: ')))
    media = media + notas[inicio]
    inicio += 1
print(f'As notas foram: {notas}')
print(f'A media é {media}')
```

```python
soma = 0

while True:
    valor = int(input('Digite um numero para somar ou 0 para sair: '))
    if valor == 0:
        break
    soma += valor
print(soma)
```

# Estrutura de repetição  **for ... in**

```python
numero = '4589456789'
soma = 0

for c in numero:
    soma += int(c)

print(soma)
```

# Faça um programa que mostre a tabuada usando for.

```python
x = int(input('Qual numero você quer a tabuada? '))

for i in range(1,11):
    print(f'{x} X {i} = {x*i}')
```

```python
p = input('Digite nome de cidade: ')

contador_A = 0
contador_R = 0
for letra in p:
    if letra.lower() == 'a':
        contador_A += 1
    if letra.lower() == 'b':
        contador_R += 1

print(f'{p} tem:')
print(f'{contador_A} de letra A')
print(f'{contador_R} de letra R')
```
